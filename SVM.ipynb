```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
import matplotlib.pyplot as plt
```

```python
# Creating dataset
data={
    "study_hours":[2,3,5,6,7],
    "attendence":[60,65,75,80,85],
    "result":["fail","fail","pass","pass","pass"]
    
}
df=pd.DataFrame(data)
df
```

```python
x=df[["study_hours","attendence"]]  ## features
y=df["result"]                       ## target
```

```python
df["result"]=df["result"].map({"fail":0,"pass":1})
df
```

```python
model=SVC(kernel="linear")
model.fit(x,y)
```

```python
w=model.coef_[0]     ## w1,w2 
b=model.intercept_[0]
print(w)
print(b)
```

```python
new_student=pd.DataFrame({
    "study_hours":[4],
    "attendence":[70]
})
pre=model.predict(new_student)
if pre[0]=="pass":
    print("prediction:pass")
else:
    print("prediction:fail")
```

```python
x_vals=np.linspace(x["study_hours"].min()-1,x["study_hours"].max()+1,100)
                   
y_vals=-(w[0]* x_vals+ b) / w[1]                   
```

```python
plt.figure(figsize=(8, 6))

# Separate data
fail = df[df.iloc[:, 2] == 0]   # Result column
pass_ = df[df.iloc[:, 2] == 1]

# Fail points
plt.scatter(
    fail.iloc[:, 0],   # Study Hours
    fail.iloc[:, 1],   # Attendance
    color="red",
    label="Fail"
)

# Pass points
plt.scatter(
    pass_.iloc[:, 0],
    pass_.iloc[:, 1],
    color="green",
    label="Pass"
)

# Hyperplane
plt.plot(x_vals, y_vals, "k--", label="Hyperplane")

# Predicted point (NUMPY ARRAY)
plt.scatter(
    new_student.iloc[0, 0],
    new_student.iloc[0, 1],
    marker="*",
    s=250,
    label="Predicted Student"
)

plt.xlabel("Study Hours")
plt.ylabel("Attendance (%)")
plt.title("SVM Classification: Pass vs Fail")
plt.legend()
plt.grid(True)
plt.show()
```
